```python
# Program Of Expenses Tracker App
print("\n---WELCOME TO EXPENSES TRACKER APP-----")

expenses = []

# 2. Define Your Functions
def amount_expenses():
    global expenses
    amount = float(input("Enter expense amount:"))
    category = input("Enter expense Category:")
    
    expenses.append({
        "amount":amount,
        "category":category
    })
    print ("Expense added!")

def  value_expenses():
    global expenses
    total = 0

    for item in expenses:
        total = total + item["amount"]
    print(f"Total value of Expenses:{total}")

def show_expenses():
    global expenses
    if len (expenses) == 0:
        print("No Expenses Adeed yet.")
    else:
        print("\n---YOUR EXPENSES----")
        for item in expenses:
            print("Amount:", ":",item["amount"])

def total_expenses():
    global expenses
    total = 0
    for item in expenses:
        total = total + item["amount"]
    print("Total:",total)

#3. Menu Selection
while True:
    print("\n1. Amount Expenses", 
             "2. Value Expenses",
             "3. Show Expenses",
             "4. Total Expenses",
             "5.Exit")
    
    select_number = int(input("Select Your Option:"))

    if select_number == 1:
         amount_expenses()
    elif select_number == 2:
        value_expenses()
    elif select_number == 3:
         show_expenses()
    elif select_number == 4:
         total_expenses()
    elif select_number == 5:
         print("Thank you!")
         break 
    else:
         print("INVALID OPTION!")
    
```
